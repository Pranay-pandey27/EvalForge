# EvalForge experiment analysis
This notebook loads saved `summary.json` artifacts. It never contains manually entered benchmark results.

```python
from pathlib import Path
from evalforge.evaluation import load_summaries, aggregate

run_root = Path('experiments/runs/model-evaluation')
summaries = load_summaries(run_root)
aggregate(summaries)
```

```python
import pandas as pd
rows = [summary.model_dump(mode='json') for summary in summaries]
pd.json_normalize(rows)[['task_id', 'condition', 'resolved', 'iterations', 'runtime_seconds']]
```
